```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Lasso,Ridge,ElasticNet
from sklearn.metrics import r2_score, mean_absolute_error,mean_squared_error,mean_absolute_percentage_error
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv(r"C:\Users\Dhruv\Downloads\insurance.csv")
df.head(5)
```

```python
df.isnull().sum()
```

```python
df['sex']=df['sex'].replace({'male':0,'female':1})
df['smoker']=df['smoker'].replace({'yes':0,'no':1})
df['region']=df['region'].replace({'northeast':0,'northwest':1,'southeast':2,'southwest':3})
```

```python
df.head(5)
```

```python
df.columns
```

```python
X=df[['age', 'sex', 'bmi', 'children', 'smoker', 'region']]
y=df[['charges']]
```

```python
# Features Selection
X_train, X_test, y_train, y_test = train_test_split(X,
                                                    y, 
                                                    test_size=0.3, 
                                                    random_state=42)
```

```python
# Scale data
scaler=StandardScaler()
X_train_scale=scaler.fit_transform(X_train,y_train)
X_test_scale=scaler.transform(X_test)
```

```python
# model selection
lin_model=LinearRegression()
lin_model.fit(X_train_scale,y_train)
pred=lin_model.predict(X_test_scale)
print("Linear prediction:",r2_score(y_test,pred))
```

```python
print("model slope :",lin_model.coef_[0])
print("model intercept :",lin_model.intercept_)
```

```python
# model lasso
ls=Lasso(alpha=100)
ls.fit(X_train_scale,y_train)
pred_ls=ls.predict(X_test_scale)
```

```python
# model Ridge
ri=Ridge(alpha=1)
ri.fit(X_train_scale,y_train)
pred_ri=ri.predict(X_test_scale)
```

```python
# model elastic net
el=ElasticNet(alpha=1,l1_ratio=0.5)
el.fit(X_train_scale,y_train)
pred_el=el.predict(X_test_scale)
```

```python
# R2 scores
print("r2 score Linear Regression:",r2_score(y_test,pred))
print("r2 score Lasso:",r2_score(y_test,pred_ls))
print("r2 score Ridge:",r2_score(y_test,pred_ri))
print("r2 score ElasticNet:",r2_score(y_test,pred_el))
```

```python
mean_absolute_error_print = mean_absolute_error(y_test,pred)
print("mean_absolute_error :",mean_absolute_error_print)

mean_absolute_error_print = mean_squared_error(y_test,pred)
print("mean_squared_error :",mean_absolute_error_print)

mean_root_mean_squared_error_print = mean_squared_error(y_test,pred)
print("mean_root_mean_squared_error :",mean_root_mean_squared_error_print)
```
