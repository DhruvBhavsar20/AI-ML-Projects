```python
import os
import cv2
import numpy as np
import pandas as pd

dataset_path = r"C:\Users\Dhruv\Downloads\cotton leaf\Augmented Dataset"

data = []

for disease in os.listdir(dataset_path):

    folder = os.path.join(dataset_path, disease)

    if not os.path.isdir(folder):
        continue

    for image_name in os.listdir(folder):

        if image_name.lower().endswith((".jpg", ".jpeg", ".png")):

            image_path = os.path.join(folder, image_name)

            # Read image
            image = cv2.imread(image_path)

            if image is None:
                continue

            # -------------------------
            # Extract features
            # -------------------------

            red = np.mean(image[:, :, 2])
            green = np.mean(image[:, :, 1])
            blue = np.mean(image[:, :, 0])

            gray = cv2.cvtColor(
                image,
                cv2.COLOR_BGR2GRAY
            )

            texture = np.std(gray)

            edges = cv2.Canny(
                gray,
                100,
                200
            )

            edge_density = np.mean(edges > 0)

            # -------------------------
            # Add ONE row
            # -------------------------

            data.append([
                image_name,
                red,
                green,
                blue,
                texture,
                edge_density,
                disease
            ])


# Create DataFrame
df = pd.DataFrame(data, columns=[
    "image",
    "red_intensity",
    "green_intensity",
    "blue_intensity",
    "texture",
    "edge_density",
    "disease"
])
df
```

```python
df.to_csv(
    "cotton_Augmented_features.csv",
    index=False
)
```
