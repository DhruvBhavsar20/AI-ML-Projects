```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, LabelEncoder, OrdinalEncoder
```

```python
df['fever']=df['fever'].fillna(df['fever'].mean())
```

```python
df.isnull().sum()
```

```python
X_train,X_test,y_train,y_test=train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2)
```

```python
# Odinal Encoding
oe=OrdinalEncoder(categories=[['Mild','Strong']])
X_train_cough=oe.fit_transform(X_train[['cough']])
X_test_cough=oe.fit_transform(X_test[['cough']])
X_test_cough.shape
```

```python
# One hot encoding
ohe=OneHotEncoder(drop='first',sparse_output=False)
X_train_gender_city=ohe.fit_transform(X_train[['gender','city']])
X_test_gender_city=ohe.fit_transform(X_test[['gender','city']])
X_train_gender_city.shape
```

```python
X_train_transformed=np.concatenate((X_train_cough,X_train_gender_city),axis=1)
X_test_transformed=np.concatenate((X_test_cough,X_test_gender_city),axis=1)
X_train_transformed.shape
```

```python
# Easy way to put multicolinearity
```

```python
from sklearn.compose import ColumnTransformer
```

```python
ct=ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(),['fever']),
    ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('tnf3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
],remainder='passthrough')
```

```python
ct.fit_transform(X_train)
ct.fit_transform(X_test)
```
