```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv(r"C:\Users\Dhruv\Downloads\student_pass (1).csv")
df.head(20)
```

```python
df['Result']=df['Result'].replace({'Pass':1,'Fail':0})
```

```python
# feature Selection
X=df[['Study_Hours','Attendance']]
y=df['Result']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,
                                               y,
                                              test_size=0.3,
                                              random_state=42)
```

```python
# Scaler Data:
scaler= StandardScaler()
X_train_scale=scaler.fit_transform(X_train,y_train)
X_test_scale=scaler.transform(X_test)
```

```python
# Model Selection
model=KNeighborsClassifier(n_neighbors=4)
model.fit(X_train_scale,y_train)
```

```python
# Prediction
pred=model.predict(X_test_scale)
print("Prediction",pred)
```

```python
# Probablity
model_proba=model.predict_proba(X_test_scale)
print("Probablity",model_proba)
```

```python
# Accuracy
accuracy=accuracy_score(y_test,pred)
print("Accuarcy",accuracy*100)
```

```python
new_data = pd.DataFrame(
    [[5.5,68]],
    columns=["Study_Hours", "Attendance"]
)

new_data_scaled = scaler.transform(new_data)
prediction = model.predict(new_data_scaled)

print("Prediction:", prediction)
```
